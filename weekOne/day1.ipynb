```python
import os
from openai import OpenAI
from scraper import fetch_website_contents
from IPython.display import display, Markdown
```

```python
system_prompt="""
You are an assistant that analyzes the contents of a website
and provides a short, clear summary, ignoring navigation-related text.
Respond in markdown.
"""
User_prompt = """ 
Here is the content of a website. Please provide a short summary of it in markdown.
"""
```

```python
def message_data(data):
    return[
        {"role" : "system", "content" : system_prompt},
        { "role" : "user", "content": User_prompt + data}
    ]
```

```python
message_data("https://edwarddonner.com")
```

```python
ollama = "http://localhost:11434/v1"
openai = OpenAI(base_url=ollama, api_key="anything")

def Summarize(url):
    website = fetch_website_contents(url)
    response=openai.chat.completions.create(
        model="llama3.2" , 
        messages = message_data(website)
    )
    return response.choices[0].message.content
```

```python
Summarize("https://edwarddonner.com")
```

```python
def displays(url):
    summary = Summarize(url)
    display(Markdown(summary))

```

```python
displays("https://edwarddonner.com")
```
